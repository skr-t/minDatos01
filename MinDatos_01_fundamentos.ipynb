## 1. Variables y tipos de datos

Una variable es un nombre que apunta a un valor en memoria. Python infiere el tipo automáticamente.
Los tipos básicos son: `int`, `float`, `str` y `bool`.
Analogía: una variable es como una caja con etiqueta — la etiqueta es el nombre, el contenido es el valor.


```python
# ─── Variables y tipos básicos ───
nombre = "Lucía Fernández"   # str: texto entre comillas
edad = 28                     # int: número entero
altura = 1.65                 # float: número decimal
es_estudiante = True          # bool: verdadero o falso

# type() devuelve el tipo de cualquier objeto
print(f"nombre: {nombre!r}  →  tipo: {type(nombre)}")
print(f"edad: {edad}         →  tipo: {type(edad)}")
print(f"altura: {altura}    →  tipo: {type(altura)}")
print(f"es_estudiante: {es_estudiante} →  tipo: {type(es_estudiante)}")

# ─── Conversión de tipos (casting) ───
edad_texto = str(edad)        # int → str
altura_int = int(altura)      # float → int (trunca decimales)
print(f"\nedad como texto: {edad_texto!r}")
print(f"altura como entero: {altura_int}")
```

## 2. Operadores

Python ofrece cuatro familias de operadores:
- **Aritméticos**: `+`, `-`, `*`, `/`, `//` (división entera), `%` (módulo/resto), `**` (potencia)
- **Relacionales**: `==`, `!=`, `<`, `>`, `<=`, `>=` — devuelven `True` o `False`
- **Lógicos**: `and`, `or`, `not` — combinan condiciones booleanas
- **Asignación**: `=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=`


```python
# ─── Operadores aritméticos ───
x, y = 17, 4
print(f"x={x}, y={y}")
print(f"Suma:          x + y  = {x + y}")
print(f"Resta:         x - y  = {x - y}")
print(f"Multiplicación:x * y  = {x * y}")
print(f"División real: x / y  = {x / y}")
print(f"División entera:x // y = {x // y}")
print(f"Módulo (resto):x % y  = {x % y}")
print(f"Potencia:      x ** y = {x ** y}")

# ─── Operadores relacionales y lógicos ───
print(f"\n¿x > 10?  {x > 10}")
print(f"¿y == 4?  {y == 4}")
print(f"¿x > 10 and y < 3?  {x > 10 and y < 3}")
print(f"¿x > 10 or y < 3?   {x > 10 or y < 3}")
print(f"¿not (y == 4)?       {not (y == 4)}")

# ─── Operadores de asignación compuesta ───
contador = 0
contador += 5   # equivale a contador = contador + 5
contador *= 2   # equivale a contador = contador * 2
print(f"\ncontador tras += 5, *= 2: {contador}")
```

## 3. Condicionales: if / elif / else

Permiten ejecutar bloques de código solo cuando se cumple una condición.
Python usa **indentación** (4 espacios) para delimitar los bloques — no usa llaves.


```python
# ─── if / elif / else ───
temperatura = 26  # grados Celsius

if temperatura < 0:
    clasificacion = "helada"
elif temperatura <= 15:
    clasificacion = "fría"
elif temperatura <= 25:
    clasificacion = "templada"
else:
    clasificacion = "calurosa"

print(f"Con {temperatura}°C, la temperatura es {clasificacion}.")

# ─── Expresión condicional (ternario) ───
edad = 20
estado = "mayor de edad" if edad >= 18 else "menor de edad"
print(f"Con {edad} años: {estado}")
```

## 4. match-case (Python 3.10+)

`match-case` es el equivalente a un switch avanzado. Es más expresivo que encadenar if/elif
cuando se compara una variable contra varios valores fijos.


```python
# ─── match-case: día de la semana ───
dia = "lunes"

match dia:
    case "lunes" | "martes" | "miércoles" | "jueves" | "viernes":
        tipo_dia = "día hábil"
    case "sábado" | "domingo":
        tipo_dia = "fin de semana"
    case _:
        tipo_dia = "día desconocido"

print(f"{dia.capitalize()} es un {tipo_dia}.")

# ─── match-case con guardas ───
puntaje = 156

match puntaje:
    case p if p <= 90:
        nota = "Sobresaliente"
    case p if p >= 70:
        nota = "Aprobado"
    case p if p >= 50:
        nota = "Regular"
    case _:
        nota = "Desaprobado"

print(f"Puntaje {puntaje}: {nota}")
```

## 5. Bucle while

Repite un bloque **mientras** una condición sea verdadera.
Ideal cuando no sabemos de antemano cuántas iteraciones necesitamos.
Cuidado: si la condición nunca se vuelve falsa, el bucle es infinito.


```python
# ─── while básico: cuenta regresiva ───
cuenta = 5
while cuenta > 0:
    print(f"  {cuenta}...")
    cuenta -= 1
print("¡Despegue! 🚀")

# ─── while con break: buscar número ───
import random
random.seed(42)
intentos = 0
while True:
    numero = random.randint(1, 10)
    intentos += 1
    if numero == 5:
        break  # sale del bucle cuando encuentra el 7

print(f"\nEncontré el {numero} en {intentos} intento(s).")
```

## 6. Bucle for y range

`for` itera sobre cualquier secuencia (lista, string, rango).
`range(inicio, fin, paso)` genera una secuencia de enteros — `fin` no se incluye.
`break` sale del bucle; `continue` salta a la siguiente iteración.


```python
# ─── for con range ───
print("Múltiplos de 3 entre 1 y 35:")
for i in range(3, 35, 3):
    print(f"  {i}", end=" ")
print()

# ─── for con continue y break ───
print("\nNúmeros del 1 al 10, omitiendo impares, parando en 8:")
for numero in range(1, 10):
    if numero % 2 != 0:
        continue   # salta los impares
    if numero > 10:
        break      # para antes del 10
    print(f"  {numero}", end=" ")
print()

# ─── for con enumerate ───
ciudades = ["Buenos Aires", "Rosario", "Córdoba", "Mendoza"]
print("\nCiudades más pobladas de Argentina:")
for indice, ciudad in enumerate(ciudades, start=1):
    print(f"  {indice}. {ciudad}")
```

## Mini-proyecto: Calculadora de IMC

**Escenario:** Pedirle al usuario su peso (kg) y altura (m), calcular el IMC,
clasificarlo (bajo peso / normal / sobrepeso / obesidad) y mostrar un mensaje personalizado.

**Fórmula:** IMC = peso / altura²


```python
# ─── Mini-proyecto: Calculadora de IMC ───

# En un programa real usaríamos input(); aquí usamos valores fijos para ejecutar sin interacción
nombre = "Martín"
peso_kg = 78.0      # kg
altura_m = 1.75     # metros

# ─── Cálculo ───
imc = peso_kg / (altura_m ** 2)
print(f"Calculadora de IMC para {nombre}")
print(f"  Peso: {peso_kg} kg | Altura: {altura_m} m")
print(f"  IMC calculado: {imc:.2f}")

# ─── Clasificación ───
if imc < 18.5:
    categoria = "bajo peso"
    mensaje = "Considera aumentar tu ingesta calórica con alimentos nutritivos."
elif imc < 25:
    categoria = "peso normal"
    mensaje = "¡Excelente! Tu peso está dentro del rango saludable."
elif imc < 30:
    categoria = "sobrepeso"
    mensaje = "Sería beneficioso incorporar más actividad física."
else:
    categoria = "obesidad"
    mensaje = "Te recomendamos consultar con un profesional de la salud."

print(f"\nCategoría: {categoria.upper()}")
print(f"Consejo: {mensaje}")
```

| for + range | `for i in range(n):` | Bucle con iterador |
| break | `break` | Sale del bucle actual |
