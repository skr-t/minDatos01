## 1. Manejo de excepciones: try / except / else / finally

Las excepciones son errores que ocurren en tiempo de ejecución. Python permite capturarlas
y responder de manera controlada en lugar de terminar el programa abruptamente.
- `try`: código que puede fallar
- `except`: qué hacer si falla (podemos capturar tipos específicos)
- `else`: se ejecuta solo si NO hubo excepción
- `finally`: se ejecuta SIEMPRE (ideal para limpiar recursos)


```python
# ─── try/except básico ───
def dividir(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError:
        print(f"  Error: no se puede dividir por cero")
        return None
    except TypeError as e:
        print(f"  Error de tipo: {e}")
        return None
    else:
        print(f"  {a} / {b} = {resultado:.4f}")
        return resultado
    finally:
        print(f"  (operación finalizada)")

print("Caso 1: 10 / 3")
dividir(10, 3)
print("\nCaso 2: 10 / 0")
dividir(10, 0)
print("\nCaso 3: '10' / 2")
dividir("10", 2)
```

## 2. raise — lanzar excepciones propias

Con `raise` podemos lanzar excepciones cuando detectamos condiciones inválidas.
Se usa para validar datos de entrada en funciones.


```python
# ─── raise con excepciones estándar ───
def calcular_raiz(numero):
    if not isinstance(numero, (int, float)):
        raise TypeError(f"Se esperaba int o float, se recibió {type(numero).__name__}")
    if numero < 0:
        raise ValueError(f"No se puede calcular la raíz de un número negativo: {numero}")
    return numero ** 0.5

# Pruebas
for valor in [16, -4, "hola", 9]:
    try:
        resultado = calcular_raiz(valor)
        print(f"√{valor} = {resultado}")
    except (TypeError, ValueError) as e:
        print(f"Error con {valor!r}: {e}")
```

## 3. Funciones: def, *args y **kwargs

`*args` captura argumentos posicionales variables en una tupla.
`**kwargs` captura argumentos con nombre en un diccionario.
Se pueden combinar en una misma función.


```python
# ─── Función básica con valores por defecto ───
def saludar(nombre, idioma="español"):
    saludos = {"español": "Hola", "inglés": "Hello", "francés": "Bonjour"}
    saludo = saludos.get(idioma, "Hola")
    return f"{saludo}, {nombre}!"

print(saludar("Lucía"))
print(saludar("Marie", "francés"))

# ─── *args: argumentos posicionales variables ───
def calcular_estadisticas(*numeros):
    '''Calcula media, mínimo y máximo de cualquier cantidad de números.'''
    if not numeros:
        return None
    return {
        "cantidad": len(numeros),
        "media": sum(numeros) / len(numeros),
        "minimo": min(numeros),
        "maximo": max(numeros),
    }

stats = calcular_estadisticas(7, 3, 9, 1, 5, 8, 2)
print(f"\nEstadísticas: {stats}")

# ─── **kwargs: argumentos con nombre variables ───
def crear_perfil(nombre, **datos):
    '''Crea un perfil con campos dinámicos.'''
    perfil = {"nombre": nombre}
    perfil.update(datos)
    return perfil

p = crear_perfil("Carlos", edad=30, ciudad="Rosario", profesion="Ingeniero")
print(f"\nPerfil: {p}")
```

## 4. Funciones lambda

Una función lambda es una función anónima definida en una sola línea: `lambda parámetros: expresión`.
Se usa cuando necesitamos una función pequeña para pasarla como argumento (ej: key en sorted).


```python
# ─── lambda básico ───
cuadrado = lambda x: x ** 2
doble = lambda x: x * 2
suma = lambda a, b: a + b

print(f"cuadrado(7) = {cuadrado(7)}")
print(f"doble(5) = {doble(5)}")
print(f"suma(3, 4) = {suma(3, 4)}")

# ─── lambda como key en sorted ───
productos = [
    {"nombre": "manzana", "precio": 1.50},
    {"nombre": "banana", "precio": 0.75},
    {"nombre": "naranja", "precio": 2.00},
    {"nombre": "kiwi", "precio": 3.50},
]

por_precio = sorted(productos, key=lambda p: p["precio"])
print("\nProductos por precio:")
for p in por_precio:
    print(f"  {p['nombre']:<12} ${p['precio']:.2f}")

# ─── lambda con múltiples criterios ───
personas = [("Ana", 28), ("Carlos", 22), ("Belen", 28), ("Diego", 22)]
ordenadas = sorted(personas, key=lambda x: (x[1], x[0]))  # primero por edad, luego por nombre
print(f"\nPersonas ordenadas: {ordenadas}")
```

## 5. Generadores con yield

Un generador es una función que usa `yield` para retornar valores uno por uno, **bajo demanda**.
A diferencia de una lista, no almacena todos los valores en memoria — ideal para secuencias grandes o infinitas.


```python
# ─── Generador simple: Fibonacci ───
def fibonacci_gen(limite):
    '''Genera números de Fibonacci hasta el límite dado.'''
    a, b = 0, 1
    while a <= limite:
        yield a           # suspende la función y retorna el valor
        a, b = b, a + b   # retoma desde aquí en la próxima llamada

print("Fibonacci hasta 100:")
fibs = list(fibonacci_gen(100))
print(fibs)

# ─── Usar el generador en un for ───
print("\nFibonacci hasta 1000 (solo los pares):")
for n in fibonacci_gen(1000):
    if n % 2 == 0:
        print(n, end=" ")
print()

# ─── Generador con expresión ───
cuadrados_gen = (x**2 for x in range(1, 6))  # paréntesis, no corchetes
print(f"\nCuadrados (generador): {list(cuadrados_gen)}")
```

## 6. Decoradores

Un decorador es una función que toma otra función como argumento y retorna una nueva función
con comportamiento extendido. Se aplica con la sintaxis `@nombre_decorador`.
Patrón: `wrapper` captura `*args, **kwargs` para ser transparente a cualquier firma.


```python
# ─── Decorador de cronómetro ───
def cronometrar(funcion):
    '''Decorador que mide el tiempo de ejecución de cualquier función.'''
    def wrapper(*args, **kwargs):
        inicio = time.perf_counter()
        resultado = funcion(*args, **kwargs)
        fin = time.perf_counter()
        print(f"  [{funcion.__name__}] tardó {(fin - inicio)*1000:.3f} ms")
        return resultado
    return wrapper

@cronometrar
def suma_lenta(n):
    '''Suma los números del 1 al n con un bucle.'''
    total = 0
    for i in range(n):
        total += i
    return total

@cronometrar
def suma_rapida(n):
    '''Suma usando la fórmula de Gauss.'''
    return n * (n - 1) // 2

print("Comparando implementaciones para n=1_000_000:")
r1 = suma_lenta(1_000_000)
r2 = suma_rapida(1_000_000)
print(f"Resultados iguales: {r1 == r2}")
```

## 7. map, filter y reduce

- `map(func, iterable)` aplica `func` a cada elemento
- `filter(func, iterable)` conserva solo los elementos donde `func` retorna True
- `reduce(func, iterable)` acumula aplicando `func` de izquierda a derecha (requiere `functools`)


```python
# ─── map ───
precios = [100, 250, 80, 430, 175]
precios_con_iva = list(map(lambda p: round(p * 1.21, 2), precios))
print(f"Precios originales:  {precios}")
print(f"Precios con IVA 21%: {precios_con_iva}")

# ─── filter ───
nombres = ["Ana", "Beto", "Catalina", "Do", "Esperanza", "Fe", "Guadalupe"]
nombres_largos = list(filter(lambda n: len(n) > 5, nombres))
print(f"\nNombres originales: {nombres}")
print(f"Nombres largos (>5): {nombres_largos}")

# ─── reduce ───
numeros = [1, 2, 3, 4, 5]
producto = reduce(lambda a, b: a * b, numeros)
print(f"\nProducto de {numeros} = {producto}")

# ─── Combinando map + filter + lambda ───
datos = ["  Python  ", "GO", "JavaScript", "Rust", "C", "  TypeScript  "]
lenguajes_largos_limpios = list(map(
    str.upper,
    filter(lambda s: len(s.strip()) > 4, map(str.strip, datos))
))
print(f"\nLenguajes (limpios, largos, en mayúsculas): {lenguajes_largos_limpios}")
```

```python
# 🎯 EJERCICIO 3: Decorador cronometrar
import time

def cronometrar(funcion):
    def wrapper(*args, **kwargs): #porque se usa una funcion dentro de otra?
        inicio = time.time()
        resultado = funcion(*args, **kwargs)
        fin = time.time()
        print(f"[{funcion.__name__}] tardó {(fin-inicio)*1000:.2f} ms")
        return resultado
    return wrapper

@cronometrar
def esperar_un_poco():
    time.sleep(0.1)
    return "listo"

# Prueba
resultado = esperar_un_poco()
print(f"Retorno de la función: {resultado}")
```

## Mini-proyecto: Validador de datos de formulario

**Escenario:** Validar los campos de un formulario de registro:
- nombre: al menos 2 caracteres, solo letras y espacios
- email: debe contener @
- edad: entero entre 0 y 120
- telefono: exactamente 10 dígitos

Retornar un diccionario con los errores encontrados (vacío si todo es válido).


```python
# ─── Mini-proyecto: Validador de formulario ───
import re

def validar_nombre(nombre):
    if len(nombre.strip()) < 2:
        return "El nombre debe tener al menos 2 caracteres"
    if not re.match(r'^[a-zA-ZáéíóúÁÉÍÓÚñÑ ]+$', nombre):
        return "El nombre solo puede contener letras y espacios"
    return None

def validar_email(email):
    if "@" not in email:
        return "El email debe contener @"
    partes = email.split("@")
    if len(partes) != 2 or not partes[0] or not partes[1]:
        return "Formato de email inválido"
    return None

def validar_edad(edad_str):
    try:
        edad = int(edad_str)
        if not 0 <= edad <= 120:
            return "La edad debe estar entre 0 y 120"
    except ValueError:
        return "La edad debe ser un número entero"
    return None

def validar_telefono(telefono):
    digitos = re.sub(r'\D', '', telefono)  # eliminar no-dígitos
    if len(digitos) != 10:
        return "El teléfono debe tener exactamente 10 dígitos"
    return None

def validar_formulario(nombre, email, edad, telefono):
    errores = {}
    for campo, validador, valor in [
        ("nombre", validar_nombre, nombre),
        ("email", validar_email, email),
        ("edad", validar_edad, edad),
        ("telefono", validar_telefono, telefono),
    ]:
        error = validador(valor)
        if error:
            errores[campo] = error
    return errores

# ─── Pruebas ───
formularios = [
    ("María González", "maria@gmail.com", "28", "1123456789"),   # válido
    ("A", "correo-sin-arroba", "150", "12345"),                   # múltiples errores
    ("Juan Pérez", "juan@mail.com", "35", "2615551234"),          # válido
]

for nombre, email, edad, tel in formularios:
    errores = validar_formulario(nombre, email, edad, tel)
    estado = "✓ VÁLIDO" if not errores else f"✗ {len(errores)} error(es)"
    print(f"\nFormulario de {nombre!r}: {estado}")
    for campo, error in errores.items():
        print(f"  - {campo}: {error}")
```

## Desafío extra 🔴

Implementar un decorador `@reintentar(veces=3, espera=0.1)` que, si una función lanza una excepción,
la vuelve a llamar hasta `veces` veces con una pausa de `espera` segundos entre intentos.
Probar con una función que falla aleatoriamente.


```python
import random
import time

def reintentar(veces=3, espera=0.1):
    # NIVEL 1: Recibe los argumentos del decorador (veces, espera)
    def decorador(funcion):
        # NIVEL 2: Recibe la función que vamos a decorar (conexion_inestable)
        def envoltura(*args, **kwargs):
            # NIVEL 3: La lógica que se ejecuta en cada llamada
            intentos_fallidos = 0

            while intentos_fallidos < veces:
                try:
                    # Intentamos ejecutar la función original
                    return funcion(*args, **kwargs)

                except Exception as e:
                    # Si falla, sumamos un intento
                    intentos_fallidos += 1
                    print(f"⚠️ Intento {intentos_fallidos} falló: {e}")

                    # Si todavía nos quedan intentos, esperamos y repetimos el bucle
                    if intentos_fallidos < veces:
                        print(f"   Reintentando en {espera}s...")
                        time.sleep(espera)
                    else:
                        # Si ya no hay más oportunidades, lanzamos la excepción
                        print("❌ Se agotaron los intentos.")
                        raise e

        return envoltura
    return decorador

# Aplicamos el decorador con parámetros específicos
@reintentar(veces=5, espera=0.05)
def conexion_inestable():
    '''Simula una conexión que falla el 70% de las veces.'''
    if random.random() < 0.7:
        raise ConnectionError("Fallo de conexión (simulado)")
    return "✅ Conexión exitosa"

# Prueba del sistema
random.seed(42) # Fijamos la semilla para que el resultado sea reproducible
try:
    resultado = conexion_inestable()
    print(resultado)
except ConnectionError:
    print("El sistema falló definitivamente tras todos los reintentos.")
```

## Resumen de conceptos

| Concepto | Sintaxis | Descripción |
|---|---|---|
| try/except | `try: ... except Exc as e:` | Captura excepciones |
| else | `else:` (tras except) | Solo si no hubo excepción |
| finally | `finally:` | Siempre se ejecuta |
| raise | `raise ValueError("msg")` | Lanza excepción |
| *args | `def f(*args):` | Tupla de args variables |
| **kwargs | `def f(**kwargs):` | Dict de kwargs variables |
| lambda | `lambda x: x*2` | Función anónima |
| yield | `def gen(): yield x` | Generador |
| Decorador | `@mi_deco` / `def deco(f):` | Modifica función |
| map | `map(func, iterable)` | Aplica función a cada elem |
| filter | `filter(func, iterable)` | Filtra por condición |
| reduce | `reduce(func, iterable)` | Acumula resultados |
