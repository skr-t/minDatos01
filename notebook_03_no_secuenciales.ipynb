## 1. Diccionarios — CRUD

Un diccionario mapea **claves** a **valores**. Las claves deben ser inmutables (str, int, tuple).
CRUD: Create (crear), Read (leer), Update (actualizar), Delete (eliminar).
Analogía: un diccionario es como una guía telefónica — buscás por nombre (clave) y obtenés el número (valor).


```python
# ─── Crear diccionario ───
estudiante = {
    "nombre": "Camila Torres",
    "edad": 22,
    "ciudad": "Buenos Aires",
    "carrera": "Ingeniería en Sistemas",
    "promedio": 8.5
}
print("Diccionario inicial:", estudiante)

# ─── Leer (acceder a valores) ───
print(f"\nNombre: {estudiante['nombre']}")
print(f"Promedio: {estudiante['promedio']}")
print(f"get() con default: {estudiante.get('beca', 'Sin beca')}")  # no lanza error si no existe

# ─── Update (actualizar / agregar) ───
estudiante["edad"] = 23                  # actualizar clave existente
estudiante["beca"] = "Beca Nación"       # agregar nueva clave
print(f"\nTras actualizar: {estudiante}")

# ─── Delete (eliminar) ───
ciudad_eliminada = estudiante.pop("ciudad")   # elimina y retorna el valor
print(f"\nCiudad eliminada: {ciudad_eliminada!r}")
del estudiante["carrera"]                     # elimina sin retornar
print(f"Tras eliminar: {estudiante}")
```

## 2. Métodos de diccionario

Los métodos `keys()`, `values()` e `items()` retornan vistas del diccionario.
`update()` fusiona dos dicts. `setdefault()` asigna un valor solo si la clave no existe.


```python
# ─── keys(), values(), items() ───
inventario = {
    "manzanas": 50,
    "bananas": 30,
    "naranjas": 0,
    "uvas": 15,
    "peras": 8
}

print("Claves:", list(inventario.keys()))
print("Valores:", list(inventario.values()))
print("\nItems (pares clave-valor):")
for producto, cantidad in inventario.items():
    estado = "✓" if cantidad > 0 else "✗ SIN STOCK"
    print(f"  {producto:<12}: {cantidad:>3} {estado}")

# ─── update() y setdefault() ───
extras = {"kiwis": 20, "mangos": 10}
inventario.update(extras)
print(f"\nTras update: {inventario}")

inventario.setdefault("cerezas", 0)   # agrega solo si no existe
print(f"setdefault cerezas: {inventario['cerezas']}")
```

## 3. Conjuntos (set)

Un `set` es una colección **no ordenada** de elementos **únicos**. No permite duplicados.
Implementa las operaciones matemáticas de teoría de conjuntos: unión (`|`), intersección (`&`),
diferencia (`-`), diferencia simétrica (`^`).


```python
# ─── Crear sets y operaciones básicas ───
grupo_a = {"Ana", "Carlos", "Lucía", "Martín", "Sofía"}
grupo_b = {"Martín", "Diego", "Sofía", "Valentina", "Carlos"}

print(f"Grupo A: {sorted(grupo_a)}")
print(f"Grupo B: {sorted(grupo_b)}")

# ─── Operaciones de conjuntos ───
union = grupo_a | grupo_b                  # todos (sin duplicados)
interseccion = grupo_a & grupo_b           # en ambos grupos
solo_a = grupo_a - grupo_b                 # solo en A
solo_b = grupo_b - grupo_a                 # solo en B
simetrica = grupo_a ^ grupo_b             # en uno u otro, no en ambos

print(f"\nUnión (todos):              {sorted(union)}")
print(f"Intersección (en ambos):    {sorted(interseccion)}")
print(f"Solo en A:                  {sorted(solo_a)}")
print(f"Solo en B:                  {sorted(solo_b)}")
print(f"Diferencia simétrica:       {sorted(simetrica)}")

# ─── Agregar y eliminar ───
grupo_a.add("Roberto")
grupo_a.discard("Ana")     # discard no lanza error si no existe
print(f"\nGrupo A modificado: {sorted(grupo_a)}")
```

## 4. frozenset

Un `frozenset` es un conjunto **inmutable** — no se puede modificar tras su creación.
Útil como clave de diccionario o como elemento de otro set (donde sets normales no serían hashables).


```python
# ─── frozenset ───
vocales = frozenset("aeiou")
print(f"frozenset de vocales: {vocales}")

# Podemos usar frozenset como clave de diccionario
combinaciones = {
    frozenset({"Python", "Django"}): "desarrollo web",
    frozenset({"Python", "NumPy", "Pandas"}): "ciencia de datos",
    frozenset({"Python", "FastAPI"}): "APIs REST",
}

busqueda = frozenset({"Python", "NumPy", "Pandas"})
print(f"\n{busqueda} → {combinaciones[busqueda]}")

# ─── Intentar modificar un frozenset ───
try:
    vocales.add("y")
except AttributeError as e:
    print(f"\nError esperado: frozenset es inmutable ({type(e).__name__})")
```

## 5. Dict Comprehension y Set Comprehension

La comprensión de diccionarios y conjuntos sigue la misma lógica que la de listas.
Dict: `{clave: valor for item in iterable if condición}`
Set: `{expresión for item in iterable if condición}`


```python
# ─── Dict comprehension ───
calificaciones = {
    "Lucía": 9.2, "Carlos": 6.8, "Ana": 7.5,
    "Diego": 4.1, "Sofía": 8.0, "Martín": 5.9
}

# Sólo aprobados (nota >= 6)
aprobados = {nombre: nota for nombre, nota in calificaciones.items() if nota >= 6}
print(f"Aprobados: {aprobados}")

# Convertir notas a escalas de 0-10 → 0-100
escala_100 = {nombre: nota * 10 for nombre, nota in calificaciones.items()}
print(f"En escala 0-100: {escala_100}")

# ─── Set comprehension ───
palabras = ["Python", "es", "un", "lenguaje", "de", "programación", "poderoso", "y", "elegante"]
palabras_largas = {p.lower() for p in palabras if len(p) > 4}
print(f"\nPalabras largas (set): {palabras_largas}")
```

Dadas dos listas de emails, encontrar: solo en lista1, solo en lista2, en ambas.


```python
# 🎯 EJERCICIO 2: Operaciones con sets de emails
lista1 = ["ana@mail.com", "carlos@mail.com", "lucia@mail.com", "martin@mail.com","ana@mail.com" ]
lista2 = ["martin@mail.com", "diego@mail.com", "lucia@mail.com", "sofia@mail.com"]
# Tu código aquí (convertí las listas a sets primero)

print(lista1)
print(set(lista1))

# Resultado esperado:
# Solo en lista1: {'ana@mail.com', 'carlos@mail.com'}
diferencia=set(lista1)-set(lista2)
print("Solo en lista1:", diferencia)

# Solo en lista2: {'diego@mail.com', 'sofia@mail.com'}
diferencia2=set(lista2)-set(lista1)
print("Solo en lista2:", diferencia2)

# En ambas: {'lucia@mail.com', 'martin@mail.com'}
compartidas = set(lista1) & set(lista2)
print("En ambas:", compartidas)

union = set(lista1) | set(lista2)
print(union)
```

## Mini-proyecto: Sistema de inventario

**Escenario:** Gestionar el stock de una tienda con funciones para agregar, actualizar y
eliminar productos. Mostrar una alerta para productos con stock bajo (< 5 unidades).


```python
# ─── Mini-proyecto: Sistema de inventario ───

inventario = {
    "leche": 12,
    "pan": 3,
    "arroz": 8,
    "aceite": 2,
    "fideos": 15,
    "sal": 4,
    "azúcar": 1,
}

def agregar_producto(inv, producto, cantidad):
    '''Agrega un nuevo producto o suma stock si ya existe.'''
    inv[producto] = inv.get(producto, 0) + cantidad
    print(f"  + {producto}: stock ahora = {inv[producto]}")

def actualizar_stock(inv, producto, nueva_cantidad):
    '''Actualiza el stock de un producto existente.'''
    if producto in inv:
        inv[producto] = nueva_cantidad
        print(f"  ~ {producto}: actualizado a {nueva_cantidad}")
    else:
        print(f"  ✗ {producto} no existe en el inventario")

def eliminar_producto(inv, producto):
    '''Elimina un producto del inventario.'''
    if producto in inv:
        del inv[producto]
        print(f"  - {producto}: eliminado")
    else:
        print(f"  ✗ {producto} no encontrado")

def mostrar_alertas(inv, umbral=5):
    '''Muestra productos con stock por debajo del umbral.'''
    alertas = {p: c for p, c in inv.items() if c < umbral}
    if alertas:
        print(f"\n⚠️  ALERTA DE STOCK BAJO (< {umbral} unidades):")
        for producto, cantidad in sorted(alertas.items(), key=lambda x: x[1]):
            print(f"     {producto:<12}: {cantidad} unidad(es)")
    else:
        print("  ✓ Todos los productos tienen stock suficiente.")

# ─── Operaciones ───
print("=== Sistema de Inventario ===")
agregar_producto(inventario, "harina", 20)
actualizar_stock(inventario, "leche", 2)
eliminar_producto(inventario, "fideos")
mostrar_alertas(inventario)

print("\nInventario final:")
for producto, cantidad in sorted(inventario.items()):
    print(f"  {producto:<12}: {cantidad}")
```

## Desafío extra 🔴

Implementar un sistema de **índice invertido** para búsqueda de texto:
- Dado un diccionario de documentos `{doc_id: texto}`, construir un índice donde
  cada palabra mapea al conjunto de documentos que la contienen.
- Función `buscar(indice, *palabras)` que retorne los documentos que contienen TODAS las palabras.
- Probar con al menos 5 documentos y búsquedas de 2-3 palabras.


```python
# 🔴 Desafío: Índice invertido para búsqueda
documentos = {
    1: "Python es un lenguaje de programación orientado a objetos",
    2: "La programación en Python permite crear aplicaciones web y de ciencia de datos",
    3: "Los objetos en Python tienen atributos y métodos",
    4: "El aprendizaje automático usa Python y sus bibliotecas como NumPy y Pandas",
    5: "Python es ideal para la automatización y el análisis de datos",
}

#construir el índice invertido y la función de búsqueda
def construir_indice(documentos):
    indice = {}
    for doc_id texto in documentos.items():
        for palabra in texto.split():
            if palabra not in indice:
                indice[palabra] = set()
            indice[palabra].add(doc_id)
    return indice
print(construir_indice(documentos))
```

## Resumen de conceptos

| Concepto | Ejemplo | Descripción |
|---|---|---|
| Dict literal | `{"k": v}` | Crear diccionario |
| Dict get | `d.get("k", default)` | Leer sin KeyError |
| Dict items | `d.items()` | Pares (clave, valor) |
| Dict comprehension | `{k: v for k, v in d.items()}` | Crear dict compacto |
| set | `{1, 2, 3}` | Conjunto sin duplicados |
| Unión | `a | b` | Todos los elementos |
| Intersección | `a & b` | Elementos comunes |
| Diferencia | `a - b` | En a pero no en b |
| Dif. simétrica | `a ^ b` | En uno pero no en ambos |
| frozenset | `frozenset({1,2,3})` | Conjunto inmutable |
