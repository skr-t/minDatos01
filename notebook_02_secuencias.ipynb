## 1. Strings y sus métodos

Un string es una secuencia inmutable de caracteres. Python incluye más de 40 métodos de string.
Los más usados son: `upper()`, `lower()`, `strip()`, `split()`, `join()`, `replace()`, `find()`, `count()`.
Analogía: un string es como una cinta de letras — podés leer cada letra por posición pero no reemplazarla in-place.


```python
# ─── Métodos básicos de string ───
frase = "  Hola Mundo Python  "

print(f"Original:   {frase!r}")
print(f"strip():    {frase.strip()!r}")       # elimina espacios extremos
print(f"upper():    {frase.strip().upper()!r}") # cada letra en mayúscula
print(f"lower():    {frase.strip().lower()!r}")
print(f"title():    {frase.strip().title()!r}")  # cada palabra en mayúscula

# ─── replace, find, count ───
texto = "Python es genial. Python es poderoso."
print(f"\nOriginal: {texto}")
print(f"replace:  {texto.replace('Python', 'Programar')}")
print(f"find:     posición de 'genial' = {texto.find('genial')}")
print(f"count:    ocurrencias de 'Python' = {texto.count('Python')}")

# ─── split y join ───
csv_line = "Ana García,30,Buenos Aires,ana@email.com"
campos = csv_line.split(",")
print(f"\nsplit: {campos}")
reunido = " | ".join(campos)
print(f"join:  {reunido}")
```

## 2. f-strings (literales de cadena formateados)

Los f-strings (Python 3.6+) permiten incrustar expresiones dentro de strings con `{}`.
Soportan especificadores de formato: `:.2f` (2 decimales), `:>10` (alinear derecha), `:,` (separador de miles).


```python
# ─── f-strings básicos ───
nombre = "Valentina"
edad = 27
promedio = 8.756

print(f"Alumna: {nombre}, Edad: {edad}, Promedio: {promedio:.2f}")

# ─── Especificadores de formato ───
precio = 1234567.89
print(f"Precio: ${precio:,.2f}")           # separador de miles
print(f"Precio: ${precio:>15,.2f}")        # alineado a la derecha en 15 chars
print(f"Porcentaje: {0.876:.1%}")          # formato porcentual

# ─── Expresiones dentro de f-strings ───
notas = [7, 9, 6, 8, 10]
print(f"\nNotas: {notas}")
print(f"Promedio: {sum(notas)/len(notas):.2f}")
print(f"Mayor nota: {max(notas)}")

# ─── f-string multilinea con alineación ───
print(f"\n{'Nombre':<15} {'Edad':>5} {'Ciudad':<15}")
print(f"{'─'*35}")
print(f"{'Lucía Martínez':<15} {28:>5} {'Córdoba':<15}")
print(f"{'Carlos López':<15} {34:>5} {'Rosario':<15}")
```

## 3. Listas

Las listas son secuencias **mutables** y **ordenadas**. Son el tipo de colección más usado en Python.
Métodos clave: `append()`, `insert()`, `extend()`, `pop()`, `remove()`, `sort()`, `reverse()`, `index()`, `count()`.


```python
# ─── Crear y modificar listas ───
ciudades = ["Buenos Aires", "Córdoba", "Rosario", "Mendoza", "Tucumán"]
print(f"Original: {ciudades}")

ciudades.append("Salta")                    # agregar al final
ciudades.insert(2, "Mar del Plata")         # insertar en posición 2
print(f"Tras append + insert: {ciudades}")

# ─── Eliminar elementos ───
ciudades.remove("Tucumán")                  # elimina por valor
ultima = ciudades.pop()                     # elimina y retorna el último
segunda = ciudades.pop(1)                   # elimina por índice
print(f"Eliminados: {ultima!r} y {segunda!r}")
print(f"Lista final: {ciudades}")

# ─── Ordenar ───
numeros = [34, 12, 78, 5, 90, 23]
numeros.sort()                              # ordena in-place
print(f"\nOrdenado ascendente: {numeros}")
numeros.sort(reverse=True)
print(f"Ordenado descendente: {numeros}")

# ─── sorted() retorna nueva lista (no modifica la original) ───
nombres = ["Zoe", "Ana", "Mateo", "Belen"]
ordenados = sorted(nombres)
print(f"\nOriginal: {nombres}")
print(f"sorted():  {ordenados}")
```

## 4. Slicing y acceso por índice

El slicing extrae sub-secuencias con la sintaxis `seq[inicio:fin:paso]`.
Los índices negativos cuentan desde el final: `-1` es el último elemento.
Aplica a strings, listas y tuplas por igual.


```python
# ─── Indexing y slicing ───
frutas = ["manzana", "banana", "cereza", "durazno", "frambuesa", "kiwi"]

print(f"Lista completa: {frutas}")
print(f"Primer elemento [0]:   {frutas[0]}")
print(f"Último elemento [-1]:  {frutas[-1]}")
print(f"Elementos [1:4]:       {frutas[1:4]}")    # índices 1, 2, 3
print(f"Desde índice 3 [3:]:   {frutas[3:]}")
print(f"Hasta índice 3 [:3]:   {frutas[:3]}")
print(f"Paso 2 [::2]:          {frutas[::2]}")    # cada 2 elementos
print(f"Invertida [::-1]:      {frutas[::-1]}")

# ─── Slicing en strings ───
texto = "Aprendizaje Automático"
print(f"\nPrimeras 11 letras: {texto[:11]!r}")
print(f"Desde posición 12:  {texto[12:]!r}")
print(f"Invertido:          {texto[::-1]!r}")
```

## 5. Tuplas y Unpacking

Las tuplas son secuencias **inmutables** — no pueden modificarse tras su creación.
Se usan para datos que no deben cambiar (coordenadas, fechas, registros fijos).
El **unpacking** asigna elementos de una secuencia a variables en una sola línea.


```python
# ─── Tuplas ───
coordenadas = (-34.6037, -58.3816)   # latitud y longitud de Buenos Aires
print(f"Coordenadas: {coordenadas}")
print(f"Latitud: {coordenadas[0]}, Longitud: {coordenadas[1]}")

# Las tuplas son inmutables:
try:
    coordenadas[0] = 0  # esto lanza TypeError
except TypeError as e:
    print(f"Error esperado: {e}")

# ─── Unpacking básico ───
nombre, edad, ciudad = ("Sofía", 31, "Mendoza")
print(f"\nNombre: {nombre}, Edad: {edad}, Ciudad: {ciudad}")

# ─── Unpacking con * (captura el resto) ───
primero, *resto, ultimo = [10, 20, 30, 40, 50]
print(f"Primero: {primero}, Resto: {resto}, Último: {ultimo}")

# ─── Swap con unpacking ───
a, b = 5, 10
a, b = b, a    # intercambio en una línea (no necesita variable temporal)
print(f"\nTras swap: a={a}, b={b}")
```

## 6. List Comprehension

La comprensión de listas es una sintaxis compacta para crear listas a partir de otras secuencias.
Formato: `[expresión for elemento in iterable if condición]`
Es más legible y eficiente que un bucle for equivalente.


```python
# ─── List comprehension básica ───
numeros = list(range(1, 11))
cuadrados = [n**2 for n in numeros]
print(f"Números:   {numeros}")
print(f"Cuadrados: {cuadrados}")

# ─── Con filtro (condición) ───
pares = [n for n in numeros if n % 2 == 0]
print(f"Pares: {pares}")

# ─── Transformación de strings ───
productos = ["  manzana  ", "  banana  ", "  naranja  ", "  uva  "]
limpios = [p.strip().capitalize() for p in productos]
print(f"\nProductos limpios: {limpios}")

# ─── Comprehension anidada (aplanar lista de listas) ───
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
plana = [num for fila in matriz for num in fila]
print(f"\nMatriz aplanada: {plana}")
```

## Mini-proyecto: Procesador de contactos

**Escenario:** Tenemos una lista de strings con formato `"Nombre Apellido, edad, ciudad"`.
Parsear cada string en un diccionario, ordenar los contactos por edad, e imprimir una tabla formateada.


```python
# ─── Mini-proyecto: Procesador de contactos ───

contactos_raw = [
    "Lucía Fernández, 34, Córdoba",
    "Martín Gómez, 28, Buenos Aires",
    "Valentina Cruz, 22, Rosario",
    "Diego Herrera, 45, Mendoza",
    "Sofía Ramírez, 31, Salta",
]

# ─── Paso 1: parsear cada string en un diccionario ───
contactos = []
for linea in contactos_raw:
    partes = linea.split(", ")
    contacto = {
        "nombre": partes[0].strip(),
        "edad": int(partes[1].strip()),
        "ciudad": partes[2].strip()
    }
    contactos.append(contacto)

# ─── Paso 2: ordenar por edad (menor a mayor) ───
contactos_ordenados = sorted(contactos, key=lambda c: c["edad"])

# ─── Paso 3: imprimir tabla formateada ───
print(f"{'Nombre':<20} {'Edad':>6} {'Ciudad':<15}")
print("─" * 43)
for c in contactos_ordenados:
    print(f"{c['nombre']:<20} {c['edad']:>6} {c['ciudad']:<15}")
```

## Resumen de conceptos

| Concepto | Ejemplo | Descripción |
|---|---|---|
| str.strip() | `" hola ".strip()` | Elimina espacios extremos |

| Slicing | `lst[1:4]` | Sub-secuencia |
